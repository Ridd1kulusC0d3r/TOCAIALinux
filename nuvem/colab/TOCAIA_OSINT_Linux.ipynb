## 1. Montar a TOCAIA nesta sessão
A célula abaixo baixa o lançador `trem`, o catálogo e as ferramentas do repositório.


```python
import os, urllib.request, pathlib, subprocess, sys

REPO = 'https://raw.githubusercontent.com/Ridd1kulusC0d3r/TOCAIALinux/main'
os.environ['TOCAIA_HOME'] = '/content/tocaia'
os.environ['TOCAIA_EDICAO'] = 'nuvem'
os.environ['TOCAIA_ETICA'] = 'aceito'  # as regras aparecem na seção 2
base = pathlib.Path('/content/tocaia')
(base/'ferramentas').mkdir(parents=True, exist_ok=True)
(base/'casos/demo-bx7').mkdir(parents=True, exist_ok=True)

def baixar(rel, destino):
    destino = base/destino
    destino.parent.mkdir(parents=True, exist_ok=True)
    urllib.request.urlretrieve(f'{REPO}/{rel}', destino)
    return destino

baixar('trem/trem.py', 'trem.py')
baixar('catalogo/catalogo.json', 'catalogo.json')
baixar('ferramentas/relogio.pyz', 'ferramentas/relogio.pyz')
baixar('casos/demo-bx7/postagens.csv', 'casos/demo-bx7/postagens.csv')

def trem(*args):
    r = subprocess.run([sys.executable, str(base/'trem.py'), *args],
                       capture_output=True, text=True)
    print(r.stdout); print(r.stderr, file=sys.stderr) if r.stderr else None
    return r.returncode

print('TOCAIA montada em', base)
trem('saudacao')
```
