## 1 · Montar a TOCAIA (clona e empacota)


```python
import os, subprocess, sys
REPO='https://github.com/Ridd1kulusC0d3r/TOCAIALinux'; BRANCH='unificado'
if not os.path.exists('/content/TOCAIALinux'):
    subprocess.run(['git','clone','--depth','1','-b',BRANCH,REPO,'/content/TOCAIALinux'], check=True)
os.chdir('/content/TOCAIALinux')
os.environ['TOCAIA_HOME']='/content/TOCAIALinux'
os.environ['TOCAIA_EDICAO']='nuvem'; os.environ['TOCAIA_ETICA']='aceito'
subprocess.run(['python3','trem/trem.py','validar'], check=True)
subprocess.run(['bash','scripts/empacotar.sh'], check=True)
def trem(*a):
    r=subprocess.run([sys.executable,'trem/trem.py',*a], capture_output=True, text=True)
    print(r.stdout); 		print(r.stderr) if r.stderr.strip() else None
    return r.returncode
trem('saudacao')
```
